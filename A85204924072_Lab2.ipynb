```python
price = np.array([2.5, 4.0, 1.0, 10.0])
qty = np.array([4, 2, 10, 1])

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[2, 0],
              [1, 3]])

P = np.array([[1, 2, 3],
              [4, 5, 6]])
C = np.array([[1, 0],
              [2, 1],
              [0, 3]])

cost = price * qty  # price times quantity

bill = price @ qty  # price and qty in dot product

elem = A * B  # element wise when matrix is with 2 dimension
prod = A @ B  # matrix product

PC = P @ C  # matrix product of P and C
PC_shape = PC.shape  # shape of P @ C

print("cost", cost)
print("bill", bill)
print("elem", elem)
print("prod", prod)
print("PC", PC)
print("PC_shape", PC_shape)

```
