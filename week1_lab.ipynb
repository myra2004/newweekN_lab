```python
def greet(name):
    return f"Hello, {name}!"


def area(width, height):
    return width * height


def average(numbers):
    return sum(numbers) / len(numbers)


print(greet("Aziza"))
print(area(5, 3))
print(average([72, 65, 90]))
print(round(average([72, 65, 90]), 1))
```

```python
gradebook = {"Aziza": 88, "Bekzod": 64, "Dilnoza": 95, "Eldor": 42}


def grade(mark):
    if mark >= 85:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 55:
        return "C"
    else:
        return "F"


print("Name       Mark  Grade")
for name in gradebook:
    mark = gradebook[name]
    print(f"{name:<10} {mark:>4}  {grade(mark)}")

total = sum(gradebook.values())
print()
print("Class average:", round(total / len(gradebook), 1))
```
